```python
import pandas as pd
import numpy as np
import requests
import warnings
warnings.filterwarnings('ignore')
print("All libraries imported sucessfully")
```

```python
print("="*35)
print("Extraction")
df=pd.read_csv('messy_sales_data.csv')
print("Extracted messy data successsfully")
print("="*35)

print("="*35)
print("Transformation")
print("Missing values per column:")
print(df.isnull().sum())
deleted_row=df['customer_name'].dropna()
print("Deleted rows:",len(deleted_row))
removed_values = df[df['customer_name'].isna()]
print(removed_values)
print("="*35)

print(df.isnull().sum())
print("Before finding nulls:",df.isnull().sum().sum(),"total missing values")
df['category'].fillna('Uncategorized',inplace=True)
df['customer_name'].fillna('Unknown',inplace=True)
df['quantity'].fillna(0,inplace=True)
df['product'].fillna('Unknown',inplace=True)
print("After fixing nulls:",df.isnull().sum().sum(),"total missing values")
```
