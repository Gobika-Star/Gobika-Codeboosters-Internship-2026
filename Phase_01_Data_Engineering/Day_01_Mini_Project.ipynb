```python
import pandas as pd
```

```python
df=pd.read_csv('student_performance.csv')
print("Dataset loaded successfully")
```

```python
#task 1
#Display title
print("==================================")
print("==================================")
print("Student Data Explorer")
print("==================================")
print("==================================")

#task 2
#dataset overview
#total  number of students(consider number of rows as students)
print("\nTotal number of students:",df.shape[0])

#total number of columns
print("Total number of Columns:",df.shape[1])

#total number of departments(unique)
count_dept=df['department'].unique()
print("Number of unique department  in this section:",len(count_dept))
print(df['department'].unique())

print("\n------------------------------------------------------------------")

#Task 3
print("\nTotal number of students in each department:")
print(df['department'].value_counts())

print("\n------------------------------------------------------------------")


#Task 4
#Highest math score
highscore_math=df.loc[df['math_score'].idxmax()]
print("\nName of the student:",highscore_math['name'])
print("Department :",highscore_math['department'])
print("HIghest Math score:",highscore_math['math_score'])
lowscore_math=df.loc[df['math_score'].idxmin()]
print("\nName of the student:",lowscore_math['name'])
print("Department :",lowscore_math['department'])
print("Lowest Math score:",lowscore_math['math_score'])

math_avg=df['math_score'].mean().round(2)
print("\nAverage of  math marks:")
print(math_avg)

prog_avg=df['programming_score'].mean().round(2)
print("\nAverage of programming score:")
print(prog_avg)

attendance_avg=df['attendance_percentage'].mean().round(2)
print("\nAverage of  attendance percentage:")
print(attendance_avg)

print("\n------------------------------------------------------------------")

#Task 5
total_score=df['english_score']+df['programming_score']+df['math_score']+df['science_score']
df['total_score'] = total_score # Add total_score as a new column to the DataFrame
totalscore_avg=df.groupby('gender')['total_score'].mean().round(2)
print("\nAverage of total score grouped by gender:")
print(totalscore_avg)

print("\n------------------------------------------------------------------")

#Task 6
#Top 5 Performers
sorted_5=df.sort_values(by='total_score',ascending=False)
print("\nTop 5 students by total score:")
print(sorted_5[['name' , 'department' ,  'total_score']].head(5))

print("\n------------------------------------------------------------------")

#Task 7
low_attendance=df[df['attendance_percentage']<75]
print("\nStudents below attendance percentage 75:")
print(low_attendance[['name', 'department' ,'attendance_percentage']])

if all(attendance > 75 for attendance in df['attendance_percentage']):
    print("Excellent! All students .")


```
